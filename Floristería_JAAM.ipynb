```python
# -*- coding: utf-8 -*-
"""Floristeria JAMM.ipynb


#Caracterísiticas que tienen las flores
Tipo, cantidad, precio.
#La floristería debe tener los siguientes servicios
Elegir producto, pasarela de pago, generación de factura, salir.
"""

flores={"tipo":["Rosa","Tulipán","Girasol","Amapola","Lirio"], "cantidad":[1,17,3,8,5], "precio":[4000,80000,5000,15000,23000]}

estado=True
opcion=0
comp=[]
pagar=0
rta=""

print("Bienvenido a mi floristería.")
while estado:
  print("***MENÚ***")
  print("¿Qué deseas hacer?")
  print("1. Visualizar nuestras flores")
  print("2. Comprar")
  print("3. Ver y modificar Carrito")
  print("4. Pagar")
  print("5. Factura de compra")
  print("0. Salir")
  opcion=input("Ingrese su opción: ")
  match opcion:
    case "1":
      print("\nEstas son nuestras flores con sus respectivas características:\n")
      #Estos son los títulos para la tabla que creará zip
      print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15}")
      print("-" * 46)
      #Aquí ya es el contenido de la tabla, que va a ser según los datos del diccionario
      for tipo, cantidad, precio in zip(flores['tipo'], flores['cantidad'], flores['precio']):
        print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14}")
    case "2":
      print("\n¿Qué flores desea comprar?")
      florac = input("Ingrese el nombre de la flor: ").strip().capitalize()
      if florac in flores["tipo"]:
        n = flores["tipo"].index(florac)
        stock_disponible = flores["cantidad"][n]

        if stock_disponible > 0:
          try:
            cant_comprar = int(input(f"¿Cuántas unidades de {florac} desea comprar? (Disponibles: {stock_disponible}): "))
            if cant_comprar <= 0:
              print("\nLa cantidad debe ser mayor a 0.\n")
            elif cant_comprar <= stock_disponible:
              costo_total = cant_comprar * flores["precio"][n]
              #para actualizar el inventario
              flores["cantidad"][n] -= cant_comprar
              comp.append({"flor": florac, "cantidad": cant_comprar, "subtotal": costo_total})
              pagar += costo_total

              print(f"\nSe han agregado {cant_comprar} {florac}(s) a tu carrito\n")
            else:
              print(f"\nNo hay suficiente stock. Solo tenemos {stock_disponible} unidades.\n")
          except ValueError:
            print("\nPor favor, pida un valor válido.\n")
        else:
          print(f"\nNo quedan unidades de {florac}.\n")
      else:
        print("\nEsa flor no existe en nuestro catálogo.\n")
    case "3":
      print("\n--- CARRITO DE COMPRAS ---")
      if len(comp) == 0:
        print("Tu carrito está vacío.\n")
      else:
        for i, item in enumerate(comp):
          print(f"{i + 1}. Flor: {item['flor']} | Cantidad: {item['cantidad']} | Subtotal: ${item['subtotal']}")
        print(f"Subtotal: ${pagar}\n")

        borrar = input("¿Deseas eliminar algún producto del carrito? (Si/No): ").strip().capitalize()
        if borrar == "Si":
          try:
            indice = int(input("Ingrese el número del producto que desea eliminar: ")) - 1
            if 0 <= indice < len(comp):
              item_a_eliminar = comp[indice]
              flor_nombre = item_a_eliminar['flor']
              cant_actual = item_a_eliminar['cantidad']

              cant_eliminar = int(input(f"¿Cuántas unidades de {flor_nombre} deseas retirar? (Máximo {cant_actual}): "))
              if 0 < cant_eliminar <= cant_actual:
                #se busca la flor que se quiere borrar
                idx_inventario = flores["tipo"].index(flor_nombre)
                precio_unitario = flores["precio"][idx_inventario]
                flores["cantidad"][idx_inventario] += cant_eliminar
                descuento = cant_eliminar * precio_unitario
                pagar -= descuento

                if cant_eliminar == cant_actual:
                  comp.pop(indice)
                  print(f"\nSe ha eliminado por completo {flor_nombre} del carrito.\n")
                else:
                  comp[indice]['cantidad'] -= cant_eliminar
                  comp[indice]['subtotal'] -= descuento
                  print(f"\nSe han retirado {cant_eliminar} unidades de {flor_nombre} del carrito.\n")
              else:
                print("\nCantidad no válida.\n")
            else:
              print("\nNúmero de artículo fuera de rango.\n")
          except ValueError:
            print("\nEntrada inválida. Operación cancelada.\n")

    case "4":
      metodo=input("¿Cuál será su método de pago? [Efectivo, Tarjeta Débito, Tarjeta Crédito, PSE]: ").strip().capitalize()
      print(f"Debe pagar {pagar+(pagar*0.19)} (IVA Incluido)")
      rta=input("¿Pagar? (Si/No): ").strip().capitalize()
      if rta=="Si":
        print("\n¡Gracias por su compra!\n")
      elif rta=="No":
        print("\n¡PAGA!\n")
      else:
        print("\nIngresa una respuesta válida.\n")

    case "5":
      if rta=="Si":
          print("\n***FACTURA DE COMPRA***")
          print("Productos comprados:")
          for item in comp:
              print(f"- Flor: {item['flor']} | Cantidad: {item['cantidad']} | Subtotal: ${item['subtotal']}")
          print(f"Total pagado (IVA INCLUIDO): ${pagar+(pagar*0.19)}")
          print(f"Método de pago: {metodo}\n")
      else:
        print("\n¡Debe realizar el pago antes de generar la factura!\n")

    case "0":
      print("\n¡Hasta pronto!\n")
      estado=False
    case _:
      print("\nOpción no válida.\n")

comp
```
